## 1 · Adiós al for
Un millón de temperaturas de °C a °F: con un ciclo `for` y con NumPy. **Mídelo tú** (en el video: for 73.7 ms y NumPy 1.54 ms; el resultado depende de tu computadora).

````python
temps = np.random.default_rng(1).uniform(-10, 40, 1_000_000)
lista = temps.tolist()

def con_for():
    resultado = []
    for t in lista:
        resultado.append(t * 9 / 5 + 32)
    return resultado

t_for = min(timeit.repeat(con_for, number=3, repeat=5)) / 3
t_np = min(timeit.repeat(lambda: temps * 9 / 5 + 32, number=3, repeat=5)) / 3
print(f"for:   {t_for * 1000:.1f} ms")
print(f"NumPy: {t_np * 1000:.2f} ms")
print(f"NumPy fue {t_for / t_np:.0f} veces más rápido")
````

## 3 · ¿Qué es vectorizar?
- Un ciclo `for` visita los elementos **uno por uno**, y en cada vuelta Python hace trabajo extra.
- **Vectorizar** es escribir la operación **una sola vez** para todo el arreglo: NumPy la aplica a todos los elementos por dentro, en código compilado.
- Tú escribes la fórmula como si fuera un solo número.

## 4 · Operaciones vectorizadas
Una operación con un número se aplica a **cada** elemento. ¿Qué tipo de dato da la división?

````python
precios = np.array([10, 25, 7, 40])
print(precios + 5)
print(precios - 5)
print(precios * 2)
print(precios / 2)
print(10 - precios)
````

## 5 · Potencias, división entera y residuo
- `**` potencia · `//` división entera (cuántas veces cabe completo) · `%` residuo (lo que sobra).

````python
x = np.array([7, 8, 9, 10])
print(x ** 2)
print(x // 3)
print(x % 3)
````

Minutos → horas y minutos, para todo el arreglo:

````python
minutos = np.array([75, 130, 45])
print(minutos // 60)
print(minutos % 60)
````

⚠️ **Cuidado con los negativos:** `//` redondea **hacia abajo**. Predice antes de ejecutar.

````python
print(np.array([-7]) // 2)
print(np.array([-7]) % 2)
````

## 6 · Combinar arreglos elemento a elemento
NumPy empareja los elementos **por posición**.

````python
precios = np.array([10, 25, 7])
cantidades = np.array([3, 1, 5])
print(precios * cantidades)
print((precios * cantidades).sum())
````

Compáralo con las listas:

````python
print([1, 2, 3] + [10, 20, 30])
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print(a + b)
````

Los dos arreglos deben tener el **mismo tamaño** (este error es a propósito):

````python
np.array([1, 2, 3]) + np.array([1, 2])
````

## 7 · Encadenar operaciones
NumPy respeta el orden de siempre: primero `*` y `/`, al final `+` y `-`.

````python
temps_c = np.array([21.5, 23.0, 19.8, 25.1])
print(temps_c * 9 / 5 + 32)
````

⚠️ **Cuidado con los paréntesis:** cambian todo y NumPy no avisa.

````python
print(temps_c * (9 / 5 + 32))
````

Y puedes seguir la cadena con métodos:

````python
print((precios * cantidades * 1.16).round(1))
print((precios * cantidades * 1.16).round(1).sum())
````

## 8 · Funciones matemáticas vectorizadas (*ufuncs*)

````python
print(np.sqrt(np.array([1, 4, 9, 16])))
print(np.abs(np.array([-3, 2, -1])))
print(np.exp(np.array([0, 1, 2])).round(2))
print(np.log10(np.array([1, 10, 100])))
````

⚠️ Las funciones de `math` trabajan con **un solo número** (error a propósito). Usa las de NumPy:

````python
math.sqrt(np.array([1, 4, 9]))
````

## 10 · Plus: Ciencia de Datos
Antes de entrenar un modelo, los datos se ponen en la misma escala.

````python
alt = np.array([150, 160, 170, 180, 190])

# Normalización min-max: todo entre 0 y 1
print((alt - alt.min()) / (alt.max() - alt.min()))

# Estandarización: media 0 y desviación 1
print(((alt - alt.mean()) / alt.std()).round(2))
````
