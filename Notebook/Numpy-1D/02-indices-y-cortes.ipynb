## 3 · Tres formas de leer un arreglo
Los corchetes aceptan:
1. **Un número** (índice) → un solo elemento.
2. **Un corte** `inicio:fin:paso` → un rango seguido.
3. **Una lista de posiciones** → elementos sueltos (*fancy indexing*).

## 4 · Índices de un arreglo
Las posiciones empiezan en **0**. Los negativos cuentan desde el final: `-1` es el último.

````python
temps = np.array([21, 23, 19, 25, 22, 18, 24])
print(temps[0], temps[3], temps[-1])
````

Con el mismo índice puedes **cambiar** un valor:

````python
temps[2] = 30
print(temps)
````

⚠️ **Cuidado:** el arreglo es de enteros (`int64`). ¿Qué crees que se guarda si asignas 21.9?

````python
temps[0] = 21.9
print(temps[0])
````

Y una posición que no existe da error (es normal: léelo con calma).

````python
temps[10]   # error a propósito: léelo
````

## 5 · Modificar con cortes
`inicio:fin` — como en `range`, **el fin no se incluye**. Si omites el inicio empieza en 0; si omites el fin llega al último.

````python
temps = np.array([21, 23, 19, 25, 22, 18, 24])
print(temps[1:4])
print(temps[:3])
print(temps[4:])
print(temps[::2])
print(temps[::-1])
````

A un corte le puedes **asignar**: un solo valor llena todas sus posiciones, o una lista del mismo tamaño.

````python
temps[1:4] = 0
print(temps)
temps[-2:] = [100, 200]
print(temps)
````

### Vista vs copia (la trampa del inicio)
Un corte **no es una copia**: es una **vista** que mira la misma memoria del arreglo original.

````python
temps = np.array([21, 23, 19, 25, 22, 18, 24])
semana = temps[0:3]   # vista
semana[0] = 99
print(temps)
````

Si quieres trabajar aparte, usa `.copy()`:

````python
temps = np.array([21, 23, 19, 25, 22, 18, 24])
copia = temps[0:3].copy()
copia[0] = 99
print(temps)
print(copia)
````

````python
# ¿Comparten memoria? Compruébalo
print(np.shares_memory(temps, temps[0:3]))
print(np.shares_memory(temps, temps[[0, 1, 2]]))
````

## 6 · Elementos dispersos (*fancy indexing*)
Le pasas una **lista de posiciones**: en cualquier orden y hasta repetidas.

````python
temps = np.array([21, 23, 19, 25, 22, 18, 24])
idx = [0, 2, 5]
print(temps[idx])
print(temps[[6, 0, 6]])
````

Esto **sí crea una copia**: cambiar la selección no toca el original…

````python
sel = temps[idx]
sel[0] = 500
print(temps)
````

…pero si **asignas directo**, cambias justo esas posiciones:

````python
temps[idx] = -1
print(temps)
````

## 8 · Plus: Ciencia de Datos
Antes de entrenar un modelo se separan los datos en **entrenamiento** y **prueba**: se mezclan las posiciones con `permutation` y se cortan.

````python
rng = np.random.default_rng(0)
idx = rng.permutation(10)
entrenar = idx[:8]
prueba = idx[8:]
print(idx)
print(entrenar, prueba)
````

En **pandas**, `iloc` usa estos mismos índices, cortes y listas de posiciones:

````python
import pandas as pd

df = pd.DataFrame({"temp": [21, 23, 19, 25, 22]})
print(df.iloc[1:4])
print(df.iloc[[0, 4]])
````

## Resumen
1. **Índice** `a[i]` → un elemento (desde 0; `-1` es el último).
2. **Corte** `a[inicio:fin:paso]` → un rango seguido; es una **vista** (cambia el original). Usa `.copy()` para trabajar aparte.
3. **Lista de posiciones** `a[[i, j, k]]` → elementos sueltos; crea una **copia** (pero `a[[i, j]] = v` sí modifica).
