## 2 · ¿Qué es NumPy?
**NumPy** = *Numerical Python*: la librería para trabajar con muchos números a la vez.

- Una **lista** de Python guarda cada número como un **objeto suelto**, en distintos lugares de la memoria.
- Un **arreglo** de NumPy los guarda **juntos**, uno detrás de otro, y **todos del mismo tipo**.

Por eso es más rápido, y por eso **pandas**, **Matplotlib** y **scikit-learn** están construidos encima de NumPy.

## 3 · Tu primer arreglo
`np.array` convierte una lista en un arreglo. Fíjate en cómo se imprime: **sin comas**.

````python
temperaturas = [21, 23, 19, 25, 22]
arr = np.array(temperaturas)
print(arr)
````

- `type` → qué objeto es (`ndarray` = arreglo de *n* dimensiones).
- `shape` → la forma; `ndim` → cuántas dimensiones; `size` → cuántos elementos.
- Los índices funcionan como en las listas: `0` es el primero, `-1` el último.

````python
print(type(arr))
print(arr.shape, arr.ndim, arr.size)
print(arr[0], arr[-1])
````

## 4 · Arreglos vs listas
### Diferencia 1: las operaciones son vectorizadas
Con una lista, `+` **une** listas. Con un arreglo, **suma el número a cada elemento**, sin ciclo `for`.

````python
lista = [1, 2, 3]
arr = np.array([1, 2, 3])
print(lista + [10])
print(arr + 10)
````

### Diferencia 2: un arreglo guarda un solo tipo
¿Qué pasa si mezclas números y texto? Predice antes de ejecutar.

````python
print(np.array([1, "dos", 3.0]))
````

### Diferencia 3: velocidad y memoria
Mídelo tú. **El resultado depende de tu computadora** (o del servidor de Colab). En el video, NumPy fue unas 12 veces más rápido (5.09 ms contra 0.44 ms).

````python
import timeit, sys

n = 1_000_000
lista = list(range(n))
arr = np.arange(n)

t_lista = min(timeit.repeat(lambda: sum(lista), number=5, repeat=5)) / 5
t_numpy = min(timeit.repeat(lambda: arr.sum(), number=5, repeat=5)) / 5
print(f"Lista: {t_lista * 1000:.2f} ms")
print(f"NumPy: {t_numpy * 1000:.2f} ms")
print(f"NumPy fue {t_lista / t_numpy:.0f} veces más rápido")

mem_lista = sys.getsizeof(lista) + sum(sys.getsizeof(x) for x in lista)
print(f"Memoria lista: ~{mem_lista / 1e6:.0f} MB · arreglo: {arr.nbytes / 1e6:.0f} MB")
````

## 5 · Tipos de datos (`dtype`)
Cada arreglo tiene un `dtype`: el tipo de **todos** sus elementos. El número del tipo son **bits** (64 bits = 8 bytes).

````python
a = np.array([1, 2, 3])
b = np.array([1.5, 2, 3])
c = np.array([True, False, True])
d = np.array(["sol", "luna"])
print(a.dtype, b.dtype, c.dtype, d.dtype)
print(b)
print(a.itemsize, a.nbytes)
````

- Con un solo decimal, **todo** el arreglo pasa a `float64`.
- `<U4` = texto (Unicode) de hasta 4 caracteres.
- `itemsize` = bytes por elemento · `nbytes` = bytes del arreglo completo (3 × 8 = 24).

## 6 · Cambiar el tipo (`astype`)
`astype` **no modifica** el arreglo: crea uno nuevo. ¿Redondea o corta? Averígualo.

````python
precios = np.array([19.99, 5.50, 3.75])
print(precios.astype(int))
print(np.round(precios).astype(int))

textos = np.array(["10", "20", "30"])
print(textos.astype(int) + 1)
````

⚠️ **Cuidado:** si el tipo es muy chico, el número no cabe (`int8` va de -128 a 127)… y NumPy **no avisa**.

````python
x = np.array([100, 200, 300], dtype=np.int16)
print(x.astype(np.int8))
````

## 7 · Funciones fábrica
NumPy puede fabricar arreglos por ti. Fíjate en la diferencia entre `arange` (inicio, fin, **paso**; el fin no se incluye) y `linspace` (inicio, fin, **cuántos**; el fin sí se incluye).

````python
print(np.zeros(5))
print(np.ones(3))
print(np.full(4, 7))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))

rng = np.random.default_rng(42)   # con semilla: siempre los mismos números
print(rng.integers(1, 7, size=10))   # 10 tiradas de un dado
````

## 8 · Graficar arreglos

````python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
plt.plot(x, y)
plt.show()
````

````python
rng = np.random.default_rng(7)
alturas = rng.normal(170, 8, size=1000)   # 1,000 estaturas en cm
plt.hist(alturas, bins=20)
plt.show()
````

## 10 · Plus: Ciencia de Datos
Cada columna de una tabla de **pandas** es, por dentro, un arreglo de NumPy con su propio `dtype`. Con millones de filas, elegir bien el tipo ahorra mucha memoria.

````python
import pandas as pd

df = pd.DataFrame({"precio": [10, 25, 7], "cantidad": [3, 1, 5]})
print(type(df["cantidad"].to_numpy()), df["cantidad"].dtype)

print(np.zeros(1_000_000, dtype=np.int64).nbytes)
print(np.zeros(1_000_000, dtype=np.int8).nbytes)
````
