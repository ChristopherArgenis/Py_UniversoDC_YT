## 1 · La pregunta del video
¿De verdad NumPy es mucho más rápido que una lista? Mídelo tú. El número exacto depende de tu computadora (o del servidor de Colab), así que **anota tu resultado**.

````python
n = 1_000_000
lista = list(range(n))
arreglo = np.arange(n)

t0 = time.perf_counter()
sum(lista)
t_lista = time.perf_counter() - t0

t0 = time.perf_counter()
arreglo.sum()
t_numpy = time.perf_counter() - t0

print(f"Lista: {t_lista:.4f} s")
print(f"NumPy: {t_numpy:.4f} s")
print(f"NumPy fue {t_lista / t_numpy:.0f} veces más rápido")
````

## 2 · Teoría
- Una **lista** de Python guarda *referencias* a objetos sueltos, cada uno en su lugar de la memoria. Pueden ser de tipos distintos.
- Un **arreglo** de NumPy guarda todos los números **juntos** y **del mismo tipo** (`dtype`) en un solo bloque. Por eso puede operar con todos a la vez, en código compilado.

````python
lista_mixta = [3, "hola", 2.5]   # una lista acepta de todo
print(lista_mixta)

a = np.array([3, 8, 1, 9])
print(a.dtype)   # ¿qué tipo crees que tiene?
````

## 3 · Concepto
Un arreglo tiene una **forma** (`shape`): cuántos elementos hay en cada dimensión.
- `(6,)` → 1D, una fila
- `(3, 6)` → 2D, una tabla
- `(3, 3, 6)` → 3D, un cubo de datos

````python
fila = np.arange(6)
tabla = np.arange(18).reshape(3, 6)
cubo = np.arange(54).reshape(3, 3, 6)

print(fila.shape, fila.ndim)
print(tabla.shape, tabla.ndim)
print(cubo.shape, cubo.ndim)
````

## 5 · Tres ejemplos de la vida real
### Ejemplo 1 · Industria: sensores de una fábrica
Temperaturas de una máquina (°C). ¿En qué lecturas pasó de 90 °C? _(Ejemplo simplificado.)_

````python
temp = np.array([71, 74, 88, 93, 79, 95, 82])
print(temp > 90)
print(temp[temp > 90])
````

### Ejemplo 2 · Ciencia: una imagen de telescopio
Una imagen en escala de grises es un arreglo 2D: cada número es el brillo de un pixel (0 a 255). _(Ejemplo simplificado.)_

````python
rng = np.random.default_rng(42)
img = rng.integers(0, 256, size=(4, 6))
print(img)
print("forma:", img.shape)
print("brillo medio:", img.mean())
````

### Ejemplo 3 · Día a día: una canción
El audio es un arreglo con miles de muestras por segundo. Bajar el volumen a la mitad = multiplicar por 0.5. _(Ejemplo simplificado.)_

````python
muestras_por_seg = 44_100
t = np.linspace(0, 1, muestras_por_seg)
audio = np.sin(2 * np.pi * 440 * t)   # nota La (440 Hz)
mas_bajo = audio * 0.5
print(len(audio), "muestras")
print(audio.max(), mas_bajo.max())
````

## 6 · Plus: Ciencia de Datos
Cada columna de una tabla de **pandas** es, por dentro, un arreglo de NumPy.

````python
import pandas as pd

df = pd.DataFrame({"producto": ["A", "B", "C", "D"], "precio": [10, 25, 7, 40]})
col = df["precio"].to_numpy()
print(type(col))
print(col * 2)
````
